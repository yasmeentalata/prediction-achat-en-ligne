# Prédiction d’achat en e-commerce

## Objectif du projet
Prédire si une session de navigation sur un site e-commerce aboutira à un achat, à partir du comportement de navigation.

Ce projet compare une régression logistique et un arbre de décision, en étudiant le compromis entre la détection des achats et le nombre de fausses alertes.

## Données
Le jeu de données **Online Shoppers Purchasing Intention**, disponible sur le dépôt UCI, contient **12 330 sessions de navigation** et **18 colonnes**, dont la variable cible `Revenue`, qui indique si un achat a eu lieu.

Environ **15,5 % des sessions aboutissent à un achat** : les classes sont donc déséquilibrées.

## Démarche
1. Exploration des données et vérification des valeurs manquantes.
2. Construction d’une référence prédisant systématiquement « pas d’achat ».
3. Entraînement d’une régression logistique avec et sans pondération des classes.
4. Étude de plusieurs seuils de décision sur un ensemble de validation.
5. Analyse des variables corrélées et comparaison avec un arbre de décision.
6. Bilan des résultats et des limites.

## Outils
Python, pandas, scikit-learn et Matplotlib.




## 1. Chargement et exploration des données


```python
import pandas as pd

url = (
    "https://archive.ics.uci.edu/static/public/468/"
    "online%2Bshoppers%2Bpurchasing%2Bintention%2Bdataset.zip"
)

df = pd.read_csv(url, compression="zip")
print(df.shape)

print("Nombre de lignes et de colonnes :", df.shape)
display(df.head())
print("\nAchats et non-achats :")
print(df["Revenue"].value_counts())
```

```python
print("Taux d'achat :", round(df["Revenue"].mean() * 100, 1), "%")
print("Valeurs manquantes par colonne :")
print(df.isna().sum())
print("Types des colonnes :")
print(df.dtypes)
```

```python
from sklearn.model_selection import train_test_split
from sklearn.dummy import DummyClassifier
from sklearn.metrics import accuracy_score, recall_score

colonnes = ["Administrative", "Informational", "ProductRelated",
            "BounceRates", "ExitRates", "SpecialDay"]

X = df[colonnes]       # Informations utilisées pour prédire
y = df["Revenue"]      # Résultat à prédire : achat ou non

X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.2, stratify=y, random_state=42
)

baseline = DummyClassifier(strategy="most_frequent")
baseline.fit(X_train, y_train)
predictions = baseline.predict(X_test)

print("Sessions d'entraînement :", len(X_train))
print("Sessions de test :", len(X_test))
print("Accuracy :", round(accuracy_score(y_test, predictions), 3))
print("Rappel des achats :", round(recall_score(y_test, predictions), 3))
```

## 3. Régression logistique et pondération des classes

```python
from sklearn.pipeline import make_pipeline
from sklearn.preprocessing import StandardScaler
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import accuracy_score, precision_score, recall_score, confusion_matrix

modele = make_pipeline(
    StandardScaler(),
    LogisticRegression(class_weight="balanced", max_iter=1000)
)

modele.fit(X_train, y_train)
predictions_modele = modele.predict(X_test)

print("Accuracy :", round(accuracy_score(y_test, predictions_modele), 3))
print("Précision des achats :", round(precision_score(y_test, predictions_modele), 3))
print("Rappel des achats :", round(recall_score(y_test, predictions_modele), 3))
print("Matrice de confusion :")
print(confusion_matrix(y_test, predictions_modele))
```

Donc finalement la pondération des classes augmente le rappel des achats, mais la précision reste faible. Il faut étudier ce compromis avant de choisir notre modèle.


```python
modele_sans_ponderation = make_pipeline(
    StandardScaler(),
    LogisticRegression(max_iter=1000)
)

modele_sans_ponderation.fit(X_train, y_train)
predictions_sans_ponderation = modele_sans_ponderation.predict(X_test)

print("Précision des achats :", round(precision_score(y_test, predictions_sans_ponderation, zero_division=0), 3))
print("Rappel des achats :", round(recall_score(y_test, predictions_sans_ponderation), 3))
print(confusion_matrix(y_test, predictions_sans_ponderation))
```

Sans pondération, le modèle annonce très rarement un achat et en manque 380 sur 382. Avec class_weight="balanced", il en détecte 270, mais produit beaucoup plus de fausses alertes. Le choix dépendra de l’importance accordée à ces deux types d’erreurs.

## 4. Exemples de prédictions

```python
probabilites = modele.predict_proba(X_test)[:, 1]

resultats = X_test.head().copy()
resultats["Achat réel"] = y_test.head().values
resultats["Probabilité prédite"] = probabilites[:5].round(3)
resultats["Achat prédit"] = predictions_modele[:5]

display(resultats)
```

## 5. Choix du seuil sur la validation et simulation des coûts

```python
from sklearn.model_selection import train_test_split
from sklearn.metrics import precision_score, recall_score, confusion_matrix

# On partage les données d'entraînement déjà créées en deux groupes.
X_apprentissage, X_validation, y_apprentissage, y_validation = train_test_split(
    X_train,
    y_train,
    test_size=0.25,
    stratify=y_train,
    random_state=42
)

# On entraîne le modèle uniquement sur le groupe « apprentissage ».
modele.fit(X_apprentissage, y_apprentissage)

# Une probabilité d'achat pour chaque visite de validation.
probabilites = modele.predict_proba(X_validation)[:, 1]

# On essaie plusieurs seuils sur les mêmes probabilités.
for seuil in [0.3, 0.5, 0.7]:
    predictions = probabilites >= seuil

    print(f"\nSeuil : {seuil}")
    print("Précision :", round(precision_score(y_validation, predictions), 3))
    print("Rappel :", round(recall_score(y_validation, predictions), 3))
    print("Matrice de confusion :")
    print(confusion_matrix(y_validation, predictions))
```

```python
# exemple fictif parmi nos trois seuils, lequel donne le coût d’erreurs le plus faible ?
cout_fp = 2 #fausse alerte
cout_fn = 10 #achat manqué

for seuil in [0.3,0.5,0.7] :
  predictions = probabilites >= seuil

#on compte les bonnes réponses et les erreurs
  vn,fp,fn, vp = confusion_matrix(y_validation, predictions,labels=[False, True]).ravel()

  cout_total = cout_fp*fp + cout_fn *fn
  print(f"Seuil : {seuil} - coût total : {cout_total } €")


```

Avec nos coûts fictifs, le seuil 0,5 donne le coût le plus faible parmi les trois testés : 2 814 €.

```python
#Comparaison à une règle simple : notre modèle apporte-t-il un avantage par rapport à répondre toujours « pas d’achat »?
cout_toujours_non = y_validation.sum() * cout_fn
cout_toujours_oui = (~y_validation).sum() * cout_fp #inverse les booléens, sa somme compte donc les non-achats

print(f"Toujours pas d'achat : {cout_toujours_non} €")
print(f"Toujours achat : {cout_toujours_oui} €")
```

Ici, 2 814 € est inférieur à 3 810 € : avec nos hypothèses le modèle apporte un avantage, même s’il commet encore des erreurs.

## 6. Évaluation exploratoire sur le test

```python
# Seuil choisi sur la validation avec nos coûts fictifs.
seuil_choisi = 0.5

# Probabilités d'achat pour les visites du test.
probabilites_test = modele.predict_proba(X_test)[:, 1]

# Une prédiction True ou False pour chaque visite.
predictions_test = probabilites_test >= seuil_choisi

# Comparaison avec les vrais résultats.
vn, fp, fn, vp = confusion_matrix(
    y_test, predictions_test, labels=[False, True]
).ravel()

cout_modele = fp * cout_fp + fn * cout_fn
cout_toujours_non = y_test.sum() * cout_fn

print("Matrice de confusion :")
print(confusion_matrix(y_test, predictions_test, labels=[False, True]))

print("Précision :", round(precision_score(y_test, predictions_test), 3))
print("Rappel :", round(recall_score(y_test, predictions_test), 3))

print(f"Coût du modèle : {cout_modele} €")
print(f"Coût si on prédit toujours pas d'achat : {cout_toujours_non} €")
```

Le seuil 0,5 a été choisi parmi trois seuils sur la validation, avec un coût fictif de 2 € par faux positif et 10 € par faux négatif.
Sur le test, le modèle repère 70,4 % des achats et réduit le coût simulé de 19,5 % par rapport à toujours prédire « pas d’achat ».
Sa limite principale est le nombre élevé de fausses alertes. Le test ayant déjà été consulté lors des premiers essais, cette évaluation reste exploratoire.

```python
import matplotlib.pyplot as plt
from sklearn.metrics import ConfusionMatrixDisplay

ConfusionMatrixDisplay.from_predictions(
    y_test,
    predictions_test,
    labels=[False, True],
    display_labels=["Pas d'achat", "Achat"],
    cmap="Blues",
    colorbar=False
)

plt.title("Résultats sur le test — seuil de 0,5")
plt.xlabel("Prédiction du modèle")
plt.ylabel("Résultat réel")
plt.show()
```

## 7. Interprétation des coefficients et analyse des corrélations

```python
regression = modele.named_steps["logisticregression"]

coefficients = pd.DataFrame({
    "Variable": X_apprentissage.columns,
    "Coefficient": regression.coef_[0]
})

display(coefficients.sort_values("Coefficient", ascending=False))
```

```python
correlations = X_apprentissage.corr()
display(correlations.round(2))
```

La corrélation entre BounceRates et ExitRates vaut 0,91, ce qui indique une forte relation linéaire positive : quand l’un est élevé, l’autre tend aussi à l’être.

```python
# Créer les tableaux sans la colonne BounceRates.

X_apprentissage_reduit = X_apprentissage.drop(columns =["BounceRates"] )
X_validation_reduit = X_validation.drop(columns = ["BounceRates"])

# Créer et entraîner un nouveau modèle.

modele_reduit = make_pipeline(StandardScaler(), LogisticRegression(class_weight = "balanced", max_iter = 1000))
modele_reduit.fit(X_apprentissage_reduit, y_apprentissage)

# Prédire sur la validation avec le seuil 0,5.
probabilities_reduit = modele_reduit.predict_proba(X_validation_reduit)[:, 1]
predictions_reduit = probabilities_reduit >= 0.5

# Calculer les erreurs et leur coût fictif.
vn,fp,fn,vp = confusion_matrix(y_validation, predictions_reduit,  labels=[False, True]).ravel()
print("Précision :", round(precision_score(y_validation, predictions_reduit), 3))
print("Rappel :", round(recall_score(y_validation, predictions_reduit), 3))
print("Coût simulé :", fp * 2 + fn * 10, "€")
```

J’ai essayé de retirer BounceRates car cette variable était très corrélée à ExitRates (0,91), ce qui peut compliquer l’interprétation des coefficients. Après avoir entraîné un modèle sans BounceRates, j’ai comparé les résultats sur la même validation, au seuil de 0,5. La précision et le rappel restent presque identiques, et le coût simulé augmente légèrement, de 2 814 € à 2 826 €. Cela suggère que BounceRates apporte peu d’information supplémentaire dans cette expérience. Je la conserve pour cette première version, mais une version simplifiée sans cette variable serait aussi envisageable.


## 8. Comparaison avec un arbre de décision

```python
from sklearn.tree import DecisionTreeClassifier

arbre = DecisionTreeClassifier(
    max_depth=3,
    class_weight="balanced",
    random_state=42
)

arbre.fit(X_apprentissage, y_apprentissage)
```

```python
import matplotlib.pyplot as plt
from sklearn.tree import plot_tree

plt.figure(figsize=(20, 9))

plot_tree(
    arbre,
    feature_names=list(X_apprentissage.columns),
    class_names=["Pas d'achat", "Achat"],
    filled=True,
    rounded=True,
    impurity=False,
    precision=3,
    fontsize=9
)

plt.show()
```

```python
# Calculer les scores d'achat pour les visites de validation.
probabilites_arbre = arbre.predict_proba(X_validation)[:, 1]

# Transformer les scores en prédictions.
predictions_arbre = probabilites_arbre >= 0.5

# Compter les bonnes prédictions et les erreurs.
vn, fp, fn, vp = confusion_matrix(
    y_validation, predictions_arbre, labels=[False, True]
).ravel()

print("Précision :", round(precision_score(y_validation, predictions_arbre), 3))
print("Rappel :", round(recall_score(y_validation, predictions_arbre), 3))
print("Faux positifs :", fp)
print("Faux négatifs :", fn)
print("Coût simulé :", fp * 2 + fn * 10, "€")
```

J’ai comparé un arbre de décision limité à une profondeur de 3 à la régression logistique, sur les mêmes données de validation et au seuil de 0,5.

L’arbre fait moins de fausses alertes (914 contre 1 017), mais manque davantage d’achats (99 contre 78). Sa précision est légèrement supérieure : 23,6 % contre 23 %. En revanche, son rappel est plus faible : 74 % contre 79,5 %.

Avec les coûts fictifs de 2 € par faux positif et de 10 € par faux négatif, les coûts simulés sont presque identiques : 2 818 € pour l’arbre contre 2 814 € pour la régression logistique. Cet écart de 4 € sur une seule validation ne permet pas de conclure qu’un modèle est globalement meilleur. Les deux modèles font surtout des erreurs différentes.


## 9. Conclusion et limites

### Bilan
Ce projet montre pourquoi l’accuracy seule ne suffit pas lorsque les classes sont déséquilibrées : prédire systématiquement « pas d’achat » donne une accuracy de 84,5 %, mais ne détecte aucun achat.

La pondération des classes améliore la détection des achats par la régression logistique, au prix d’un nombre important de fausses alertes.

Sur la validation, la régression logistique et l’arbre de décision obtiennent des coûts simulés très proches : respectivement 2 814 € et 2 818 €. Cet écart ne permet pas d’affirmer qu’un modèle est globalement meilleur.

### Modèle retenu
Je retiens la régression logistique pondérée avec un seuil de 0,5 pour cette première version. Elle permet de détecter davantage d’achats que l’arbre sur la validation tout en restant simple à interpréter.

Sur le test exploratoire, elle détecte **70,4 % des achats**, avec une **précision de 21,7 %**. Cela signifie qu’environ un achat prédit sur cinq correspond réellement à un achat.

Avec les coûts fictifs retenus (2 € par faux positif et 10 € par faux négatif)
le coût simulé atteint **3 076 €**, contre **3 820 €** pour une prédiction systématique « pas d’achat », soit une réduction de **19,5 %**.

### Limites
- La précision reste faible : le modèle génère beaucoup de fausses alertes.
- Les coûts sont hypothétiques : la réduction calculée ne représente pas des économies réellement mesurées.
- Le test a été consulté pendant les premières expériences : ses résultats restent exploratoires.
- L’étude utilise six variables et un seul découpage des données.
- Les scores produits par le modèle pondéré ne sont pas nécessairement des probabilités d’achat bien calibrées.
- Les variables décrivent une session de navigation : une utilisation en temps réel nécessiterait de vérifier quelles informations sont disponibles au moment de prédire.

### Pistes d’amélioration
Évaluer la stabilité des résultats par validation croisée sur les données d’entraînement, étudier d’autres variables disponibles au moment de la prédiction et définir les coûts d’erreur à partir d’un besoin métier réel.
